```python
# Write a function to display users and their connections
def display_users(data):
    print("User Information\n") # new line for better visibility
    for user in data['users']:
        ''' f"{user['id']}.{user['name']} is fri... MAKES both the values 
        to execute sumultaneously by giving the 
        since its user[id] it displaying from 1 to n and not the pages '''
        print(f"ID:{user['id']}. {user['name']} is friends with: {user['friends']} and liked pages are {user['liked_pages']}")
    print("\nPages Information") # extra new line
    for page in data['pages']:
        print(f"{page['id']}: {page['name']}")
```

```python
display_users(data)
```
