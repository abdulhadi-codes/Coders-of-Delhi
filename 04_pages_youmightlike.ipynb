```python
import json

def load_data(filename):
    with open(filename,'r') as f:
        return json.load(f)
data = load_data("data.json")
```

```python
def find_pages_umili(user_id,data):
    user_pages = {}
    # populate the dict
    for user in data['users']:
        user_pages[user['id']] = set(user['liked_pages'])
    # if no user in dict
    if user_id not in user_pages :
        return []
    # there are users then 
    user_liked_pages = user_pages[user_id]
    page_suggestion = {}
    
    # Understand : If the number of liked pages between 2 users is x, if the overlapping pages between the two is 4 
    # if the overlapping pages between 2 is 4 then we take in all the overlapping pages that is x also for pages_u_might_like
    
    
    # other user is leaving the user what is intersecting
    for other_user,pages in user_pages.items():
        if other_user != user_id:
            # user and other user like the common pages the the below
            shared_pages = user_liked_pages.intersection(pages) # pages are the pages which the other user likes 
            for page in pages:
                if page not in user_liked_pages:
                    page_suggestion[page] = page_suggestion.get(page,0) + len(shared_pages)
    sorted_pages = sorted(page_suggestion.items(),key = lambda x : x[1], reverse = True)
    return [(page_id,score) for page_id,score in sorted_pages]

data = load_data("data.json")
user_id = 3
page_recomendation = find_pages_umili(user_id,data)
print(page_recomendation)
```
