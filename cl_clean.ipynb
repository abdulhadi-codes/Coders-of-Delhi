```python
import json

def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data


def clean_data(data):
    
    # Fix empty/whitespace names — replace only if empty, strip real names
    for user in data['users']:
        if not user['name'].strip():
            user['name'] = "Unknown User"
        else:
            user['name'] = user['name'].strip()
    
    # Remove duplicate friend ids
    for user in data['users']:
        user["friends"] = list(set(user['friends']))
    
    # Remove inactive users (no friends AND no liked pages)
    data["users"] = [user for user in data['users'] if user['friends'] or user['liked_pages']]
    
    # Remove duplicate page ids, warn if names conflict
    unique_pages = {}
    for page in data['pages']:
        if page['id'] in unique_pages and unique_pages[page['id']]['name'] != page['name']:
            print(f"Warning: duplicate page id {page['id']} — '{unique_pages[page['id']]['name']}' overwritten by '{page['name']}'")
        unique_pages[page['id']] = page
    data['pages'] = list(unique_pages.values())
    
    return data


# run
data = load_data("data2.json")
data = clean_data(data)
json.dump(data, open("cleaned_data3.json", "w"), indent=4)
print("Data has been updated")
```
