```python
import json

def load_data(filename):
    with open(filename, "r") as f:
        return json.load(f)
    
def pplumykn(user_id, data):
    # Stage 1: build the full lookup — runs completely first
    user_friends = {}
    for user in data["users"]:
        user_friends[user['id']] = set(user['friends'])
    
    # Stage 2: now that user_friends is fully built, compute suggestions
    if user_id not in user_friends:
        return []
    
    direct_friends = user_friends[user_id]
    suggestions = {}
    for friend in direct_friends:
        for mutual in user_friends[friend]:
            if mutual != user_id and mutual not in direct_friends:
                suggestions[mutual] = suggestions.get(mutual, 0) + 1
    
    sorted_suggestions = sorted(suggestions.items(), key=lambda x: x[1], reverse=True)
    return [candidate_id for candidate_id, _ in sorted_suggestions]
            
    # TRY FOR BOTH OF IT 
# data = load_data("data.json")
data = load_data("massive_data.json")

user_id = 3

recc = pplumykn(user_id,data)
print(f"People you may know {user_id} : {recc}")
```
